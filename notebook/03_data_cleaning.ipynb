```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/smartphones.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
```

```python
df.info()
```

```python
missing = pd.DataFrame({
    "Kolom": df.columns,
    "Jumlah Missing": df.isnull().sum(),
    "Persentase Missing (%)": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing[missing["Jumlah Missing"] > 0])
```

```python
duplicate_count = df.duplicated().sum()

print("Jumlah data duplikat:", duplicate_count)
```

```python
numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing value setelah cleaning:")
print(df.isnull().sum())
```

```python
print("Ukuran dataset akhir:", df.shape)

display(df.head())
```

```python
output_path = "../data/processed/01_clean_smartphones.csv"

df.to_csv(
    output_path,
    index=False
)

print("Data cleaning berhasil!")
print("File disimpan di:", output_path)
```
